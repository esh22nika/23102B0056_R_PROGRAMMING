Packages

```r
installed = installed.packages()
installed[, c("Package", "Version")]
```

```r
if (!requireNamespace("spatial", quietly = TRUE)) {
  install.packages("spatial", repos = "https://cloud.r-project.org")
}
library(spatial)
packageDescription("spatial")
library(help = "spatial")
```

```r
if (!requireNamespace("boot", quietly = TRUE)) {
  install.packages("boot", repos = "https://cloud.r-project.org")
}
if (!requireNamespace("cluster", quietly = TRUE)) {
  install.packages("cluster", repos = "https://cloud.r-project.org")
}
library(boot)
library(cluster)
search()
```
