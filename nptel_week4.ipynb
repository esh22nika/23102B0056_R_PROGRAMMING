Matrix operations and logical operators

Row and column names

```r
x = matrix(data = 1:12, nrow = 4, ncol = 3)
rownames(x) = c("r1", "r2", "r3", "r4")
colnames(x) = c("c1", "c2", "c3")
x
```

Constant and diagonal matrices

```r
matrix(data = 2, nrow = 4, ncol = 2)
diag(1, nrow = 3, ncol = 3)
diag(5, nrow = 3, ncol = 3)
```

Transpose

```r
x = matrix(data = 1:8, nrow = 4, ncol = 2, byrow = TRUE)
x
t(x)
```

Row and column totals

```r
x = matrix(data = 1:8, nrow = 4, ncol = 2)
x
rowSums(x)
colSums(x)
rowMeans(x)
colMeans(x)
```

Selecting values

```r
x = matrix(data = 1:15, nrow = 5, ncol = 3, byrow = TRUE)
x
x[3, ]
x[, 2]
x[4:5, 2:3]
x[c(1, 4), c(1, 3)]
x[, 2, drop = FALSE]
```

Calculations with constants

```r
x = matrix(data = 1:8, nrow = 4, ncol = 2, byrow = TRUE)
x + 5
x - 5
5 * x
x / 2
```

Adding and subtracting matrices

```r
y = matrix(data = 11:18, nrow = 4, ncol = 2, byrow = TRUE)
x + y
x - y
x + 4 * x
4 * x - x
```

Matrix multiplication

```r
y = matrix(data = 11:18, nrow = 2, ncol = 4, byrow = TRUE)
x %*% y
y %*% x
```

```r
t(x) %*% x
x %*% t(x)
crossprod(x)
```

Joining matrices

```r
x = matrix(data = 1:6, nrow = 3, ncol = 2, byrow = TRUE)
y = matrix(data = 11:16, nrow = 3, ncol = 2, byrow = TRUE)
rbind(x, y)
cbind(x, y)
```

Inverse and eigenvalues

```r
y = matrix(c(84, 100, 100, 120), nrow = 2, byrow = TRUE)
y
solve(y)
round(y %*% solve(y), 8)
eigen(y)
```

Comparisons

```r
8 > 7
7 < 5
7 >= 7
8 <= 8
7 == 7
8 != 9
```

```r
x = c(1, 2, 3)
y = c(4, 5, 6)
x > y
x < y
x != y
x == y
```

Logical values

```r
x = TRUE
y = FALSE
x & y
x | y
!x
xor(x, y)
as.logical(c(0, 1, -2))
```

```r
x = 5
logical1 = x > 2
logical2 = x < 10
logical3 = x != 5
logical1
logical2
logical3
is.logical(logical1)
2 * x > 11
3 * x < 20
```

And, or and filtering

```r
x = 1:6
(x > 2) & (x < 5)
x[(x > 2) & (x < 5)]
(x > 2) | (x < 5)
x[(x > 2) | (x < 5)]
```

```r
x = c(8, 18)
(x < 10) | (x < 2)
(x < 10) & (x > 2)
# Use single values with && and || in current R.
(x[1] < 10) || (x[1] < 2)
(x[1] < 10) && (x[1] > 2)
```

```r
x = 5
(x < 10) && (x > 2)
x = 15
(x < 10) && (x > 2)
x = 18
(x < 10) || (x < 2)
```

Checking true and false

```r
isTRUE(8 < 6)
isTRUE(8 > 6)
isFALSE(5 < 8)
isFALSE(5 > 8)
```
