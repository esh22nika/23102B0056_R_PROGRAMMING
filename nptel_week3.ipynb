Week 3

Powers, functions and matrices

Vector addition

```r
c(2, 3, 5, 7) + c(-2, -3, -5, 8)
```

Powers

```r
2^3
2**3
2^0.5
2^-0.5
```

```r
x = c(2, 3, 5, 7)
x^2
x^c(2, 3)
c(1, 2, 3, 4, 5, 6)^c(2, 3, 4)
# This one gives a recycling warning.
x^c(2, 3, 4)
```

Integer division and remainder

```r
5 %/% 2
7 %/% 3
7 %% 3
7 %% 4
```

```r
x %/% 2
x %/% c(2, 3)
x %% 2
x %% c(2, 3)
```

```r
# Unequal recycling gives warnings here too.
c(2, 3, 5) %/% c(2, 3)
c(2, 3, 5) %% c(2, 3)
```

Maximum, minimum and mean

```r
x = c(1.2, 3.4, -7.8)
max(x)
min(x)
mean(c(2, 3, 4))
# Put the values together before finding the mean.
mean(2, 3, 4)
```

Built-in functions

```r
abs(c(-1, -2, -3, 4, 5))
sqrt(c(4, 9, 16, 25))
sum(c(2, 3, 5, 7))
prod(c(2, 3, 5, 7))
```

```r
round(c(1.23, 1.83))
round(1.2345, digits = 2)
floor(c(1.23, 1.83))
ceiling(c(1.23, 1.83))
```

```r
log(c(10, 100, 1000))
log10(c(10, 100, 1000))
log2(c(2, 4, 8))
exp(1)
log(exp(1))
```

```r
# Angles are in radians.
sin(pi / 2)
cos(0)
tan(pi / 4)
asin(1)
acos(1)
atan(1)
sinh(1)
cosh(1)
tanh(1)
asinh(1)
acosh(2)
atanh(0.5)
```

Saving calculations

```r
x1 = c(1, 2, 3, 4)
x2 = x1^2
x1
x2
x1 + sum(x1) * prod(c(1, 2))
abs(x1 - sum(x1) * prod(c(1, 2)))
```

Creating matrices

```r
# Values fill down the columns by default.
x = matrix(data = 1:8, nrow = 4, ncol = 2)
x
```

```r
y = matrix(data = 1:8, nrow = 4, ncol = 2, byrow = TRUE)
y
x[3, 2]
y[3, 2]
```

Matrix properties

```r
dim(x)
nrow(x)
ncol(x)
mode(x)
attributes(x)
is.matrix(x)
```

Converting to a matrix

```r
as.matrix(1:10)
data(warpbreaks)
is.matrix(warpbreaks)
warpbreaks[1:10, ]
as.matrix(warpbreaks[1:10, ])
```
