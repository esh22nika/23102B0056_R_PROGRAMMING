Calculations with two vectors

```r
x = c(2, 3, 5, 7)
y = c(-2, -3, -5, 8)
x + y
x - y
x * y
c(24, 20, 8, 16) / c(3, 4, 2, 8)
```

Recycling

```r
# R repeats the shorter vector.
x + c(8, 9)
c(12, 13, 15, 17) - c(8, 9)
x * c(8, 9)
c(24, 20, 8, 16) / c(4, 2)
```

```r
# These give warnings because the lengths do not divide evenly.
x + c(8, 9, 10)
c(12, 13, 15, 17) - c(8, 9, 10)
x * c(8, 9, 10)
c(24, 20, 8, 16) / c(4, 2, 8)
```
