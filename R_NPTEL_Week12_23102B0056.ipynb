```r
# remove all data
rm(list = ls())

# input data vectors
x = c(10, 20, 30)
y = c(1, 2, 3)
```

```r
example1 = function(x, y)
{
  # number of observations
  n = length(x)
  # initialise the vectors which store the squares
  x1 = 0
  y1 = 0
  z1 = 0
  # start of loop
  for (i in 1:n)
  {
    x1[i] = x[i]^2
    y1[i] = y[i]^2
    z1[i] = (x[i]/y[i])^2
  }
  # sums of the squared quantities
  sum_square_x = sum(x1)
  sum_square_y = sum(y1)
  sum_square_z = sum(z1)
  # g and h
  g = sum_square_x/sum_square_y
  h = sum_square_z
  # format the output
  cat("The value of g and h are", g, "and", h, "respectively", "\n")
}
```

```r
example1       # printing the function shows its code
example1(x, y)
```

```r
# only the data changes, the function stays the same
x = c(67, 87, 26, 85, 6, 45)
y = c(54, 64, 22, 94, 20, 88)
example1(x, y)
```

```r
# alternative approach without a loop
g = sum(x^2)/sum(y^2)
h = sum(x/y)^2         # this is how it is written on the slide
g
h
h = sum((x/y)^2)       # this one is the same as h from example1
h
```
